### 3 clusters

```python
import sklearn
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
# Generate synthetic 2D data with 3 distinct clusters

X, _ = make_blobs(n_samples=300, centers=3, n_features=2, random_state=42)

# Create KMeans model to find 3 clusters
kmeans = KMeans(n_clusters=3, init='k-means++', random_state=42)
kmeans.fit(X)

# Cluster assignments and centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# Plot data points with cluster coloring
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', marker='o', edgecolor='k')
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=200, label='Centroids')
plt.title('K-Means Clustering')
plt.legend()
plt.show()
```

## Failure Cases of K-Means Clustering

```python
def run_and_plot_kmeans(X, k):
    kmeans = KMeans(n_clusters=k, random_state=42)
    clusters = kmeans.fit_predict(X)
    centers = kmeans.cluster_centers_

    plt.figure(figsize=(8,6))
    plt.scatter(X[:, 0], X[:, 1], c=clusters, cmap='Dark2', marker='.')
    plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, marker='X', label='Centroids')
    plt.title(f'K-Means Clustering with k={k}')
    plt.legend()
    plt.axis('equal')
    plt.show()
```

#### 1. Donut Problem (Non-convex clusters)

```python
import numpy as np
def generate_donut_data():
    N = 1000
    radius_inner, radius_outer = 5, 10

    r_inner = np.random.randn(N // 2) + radius_inner
    theta_inner = 2 * np.pi * np.random.rand(N // 2)
    X_inner = np.column_stack([r_inner * np.cos(theta_inner), r_inner * np.sin(theta_inner)])

    r_outer = np.random.randn(N // 2) + radius_outer
    theta_outer = 2 * np.pi * np.random.rand(N // 2)
    X_outer = np.column_stack([r_outer * np.cos(theta_outer), r_outer * np.sin(theta_outer)])

    X = np.vstack([X_inner, X_outer])
    return X
print("Donut-shaped clusters:")
X_donut = generate_donut_data()
run_and_plot_kmeans(X_donut, k=2)
```

#### 2. Elongated or Anisotropic Clusters
- Clusters are not spherical but elongated.
- K-Means struggle because it minimizes Euclidean distance to centroids and favors compact spherical clusters.
- This leads to poor cluster assignment and distorted shapes in results.

```python
def generate_elongated_clusters():
    N = 1000

    mean1 = [0, 0]
    cov1 = [[1, 0], [0, 20]]
    cluster1 = np.random.multivariate_normal(mean1, cov1, N // 2)

    mean2 = [5, 0]
    cluster2 = np.random.multivariate_normal(mean2, cov1, N // 2)

    X = np.vstack([cluster1, cluster2])
    return X
print("Elongated clusters:")
X_elongated = generate_elongated_clusters()
run_and_plot_kmeans(X_elongated, k=2)
```

#### 3. Clusters with Different Densities
- When clusters have very different densities and sizes, K-Means can assign points inaccurately.
- It tends to "steal" points from smaller, denser clusters into larger ones, resulting in imbalanced clustering.

```python
def generate_different_density_clusters():
    N = 1000

    dense_cluster = np.random.randn(950, 2)
    sparse_cluster = np.random.randn(50, 2) + [3, 0]

    X = np.vstack([dense_cluster, sparse_cluster])
    return X

print("Clusters with different densities:")
X_density = generate_different_density_clusters()
run_and_plot_kmeans(X_density, k=2)
```
