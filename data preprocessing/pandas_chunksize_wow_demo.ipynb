# Pandas Memory Demo – The "WOW Moment"
This notebook demonstrates why large datasets should sometimes be loaded **in chunks** instead of loading the entire dataset into memory.

We will:
1. Create a large synthetic dataset
2. Load the full dataset normally
3. Load the dataset using `chunksize`
4. Compare memory usage


## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
```

## Step 2: Create a Large Dataset
We simulate a large dataset with **500,000 rows**.

```python
rows = 500000

data = pd.DataFrame({
    'age': np.random.randint(18, 60, rows),
    'salary': np.random.randint(20000, 120000, rows),
    'experience': np.random.randint(0, 20, rows)
})

data.to_csv('large_dataset.csv', index=False)

print('Dataset created with shape:', data.shape)
```

## Step 3: Load Entire Dataset Normally

```python
df = pd.read_csv('large_dataset.csv')

print('Rows loaded:', df.shape[0])
print('Memory usage (bytes):', df.memory_usage(deep=True).sum())
```

## Step 4: Load Dataset Using Chunks
Instead of loading everything at once, we process the file **10,000 rows at a time**.

```python
total_rows = 0

for chunk in pd.read_csv('large_dataset.csv', chunksize=10000):
    total_rows += chunk.shape[0]

print('Rows processed through chunks:', total_rows)
```

## Teaching Moment
Ask students:

**What happens if the dataset is 10GB?**

Normal loading may crash the system.

Chunk loading allows processing data **without exhausting RAM**.
