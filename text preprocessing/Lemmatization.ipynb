```python
data = {
    "review":[
        "The boys are running in the park",
        "She was better at solving problems",
        "Students are learning machine learning"
    ]
}

df = pd.DataFrame(data)
```

### Convert POS Tags to WordNet Format
- POS tags from nltk.pos_tag() are different from WordNet POS tags, so we convert them.

```python
def get_wordnet_pos(tag):
    
    if tag.startswith('J'):
        return wordnet.ADJ
    
    elif tag.startswith('V'):
        return wordnet.VERB
    
    elif tag.startswith('N'):
        return wordnet.NOUN
    
    elif tag.startswith('R'):
        return wordnet.ADV
    
    else:
        return wordnet.NOUN
```

```python
get_wordnet_pos("NN")
```

```python
lemm = WordNetLemmatizer()
```

## Lemmatization Function

```python
def lemmatize_text(text):
    
    tokens = nltk.word_tokenize(text)
    print("*******Tokens are : ",tokens)
    pos_tags = nltk.pos_tag(tokens)
    print("#######POS Tagging : ",pos_tags)
    lemmas = [
        lemm.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in pos_tags
    ]
    
    return " ".join(lemmas)
```

```python
lemmatize_text("The boys are running in the park")
```

```python
lemm.lemmatize("are",pos="v")
```

## Apply on Dataset

```python
df["clean_review"] = df["review"].apply(lemmatize_text)

df
```
